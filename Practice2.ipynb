```python
from scipy import stats
import numpy as np

old_method = np.array([65, 70, 68, 72, 69, 71, 66])
new_method = np.array([75, 78, 80, 76, 82, 79, 77])

t_stat, p_value = stats.ttest_ind(old_method, new_method)

print("t-statistic:", t_stat)
print("p-value:", p_value)

if p_value < 0.05:
    print("পার্থক্যটা statistically significant — নতুন পদ্ধতি সত্যিই আলাদা ফল দিচ্ছে")
else:
    print("যথেষ্ট প্রমাণ নেই, পার্থক্যটা কাকতালীয়ও হতে পারে")
```

```python
from scipy import stats

n = 10       # মোট ট্রায়াল
p = 0.5      # প্রতিবার হেড আসার সম্ভাবনা

# ঠিক ৫ বার হেড আসার সম্ভাবনা কত?
prob_5_heads = stats.binom.pmf(5, n, p)
print("ঠিক ৫ বার হেড আসার সম্ভাবনা:", prob_5_heads)
```

```python
n = 10
p = 0.25

for correct in range(0, 11):
    prob = stats.binom.pmf(correct, n, p)
    print(f"{correct}টা সঠিক হওয়ার সম্ভাবনা: {prob:.4f}")
```
