What is this saying?
df.isnull() checks every cell to see if it's empty (missing) or not — if it's missing, it returns True, otherwise False
.sum() adds up how many True values (i.e., missing values) exist in each column
What the result means

Every column shows 0 next to it — meaning:

Date column has 0 missing values
Country column has 0 missing values
Confirmed column has 0 missing values
Recovered column has 0 missing values
Deaths column has 0 missing values

In short: this dataset has no missing/empty data — the entire dataset is complete (clean).

Let's make a project. load the data now

```python
import pandas as pd

url = "https://raw.githubusercontent.com/datasets/covid-19/main/data/countries-aggregated.csv"
df = pd.read_csv(url)

print(df.shape)
print(df.isnull().sum())
print(df.duplicated().sum())
```

top 10 confired country

```python
top10 = df.groupby ("Country")["Confirmed"].max().sort_values(ascending=False).head(10)
print(top10)
```

top ten death

```python
top10 = df.groupby ("Country")["Deaths"].max().sort_values(ascending=False).head(10)
print(top10)
```

BD VS INDIA

```python
bd = df[df["Country"] == "Bangladesh"].tail(1)
india = df[df["Country"] == "India"].tail(1)
print(bd)
print(india)
```

Recover Rate

```python
latest = df.groupby("Country").max(0)
latest["Recover_Rate"] = latest["Recovered"] / latest["Confirmed"] * 100
latest.sort_values("Recover_Rate", ascending=False).head(10)
```
