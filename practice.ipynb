```python
students = {
    "Rahim": 78,
    "Karim": 92,
    "Salma": 85,
    "Nadia": 65
}

def grade_checker(score):
    if score >= 90:
        return "A+"
    elif score >= 80:
        return "A"
    else:
        return "Needs Improvement"

for name, score in students.items():
    grade = grade_checker(score)
    print(name, "-", score, "-", grade)
```

```python
class Solution:
    def twoSum(self, nums = [2,7,11,15], target = 9):
        n = len(nums)
        for i in range(n):
            for j in range(i + 1, n):
                if nums[i] + nums[j] == target:
                   return [i, j] # মান রিটার্ন করবে

# ফাংশনটি কল করে প্রিন্ট করা:

print(Solution().twoSum())
```
