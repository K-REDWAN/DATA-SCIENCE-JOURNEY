```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

url = "https://raw.githubusercontent.com/datasets/covid-19/main/data/countries-aggregated.csv"
df = pd.read_csv(url)

latest = df.groupby("Country")[["Confirmed", "Recovered", "Deaths"]].max()
sns.heatmap(latest.corr(), annot=True, cmap="coolwarm")
plt.title("COVID Correlation Heatmap")
plt.show()
```

```python
import matplotlib.pyplot as plt
cities = ["Dhaka", "Chittagong", "Sylhet", "Khulna"]
population = [22, 9, 4, 3]   # মিলিয়নে

plt.bar(cities, population, color="green")
plt.title("city wise population")
plt.xlabel("city")
plt.ylabel("population in million")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

scores = np.random.normal(70, 10, 1000)   # ১০০০ জনের নম্বর, গড় ৭০

plt.hist(scores, bins=1000, color="skyblue", edgecolor="black")
plt.title("NUMBER Distribution")
plt.xlabel("NUMBER")
plt.ylabel("HOS MANY")
plt.show()
```

```python
import matplotlib.pyplot as plt
study_hours = [1, 2, 3, 4, 5]
exam_score = [50, 55, 65, 70, 80]

plt.scatter(study_hours, exam_score, color="red")
plt.title("Time VS Number")
plt.xlabel("read hour")
plt.ylabel("number")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

url = "https://raw.githubusercontent.com/datasets/covid-19/main/data/countries-aggregated.csv"
df = pd.read_csv(url)

bd = df[df["Country"] == "Bangladesh"]
bd.plot(x="Date", y="Confirmed", figsize=(10,5))
plt.title(" Trend")
plt.show()
```

```python
import matplotlib.pyplot as plt
days = [1, 2, 3, 4, 5]
class_A = [60, 65, 70, 75, 80]
class_B = [55, 58, 62, 65, 70]

plt.plot(days, class_A, label="Class A", color="blue")
plt.plot(days, class_B, label="Class B", color="orange")
plt.legend()   # কোন লাইন কোনটা বোঝানোর জন্য
plt.title("two class averaage")
plt.show()
```
