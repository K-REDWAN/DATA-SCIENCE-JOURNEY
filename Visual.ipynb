```python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5]
study_hours = [1, 2, 2, 3, 4]
plt.plot(days, study_hours)
plt.title("days wise study")
plt.xlabel("day")
plt.ylabel("time in hour")
plt.show()
```

```python
import matplotlib.pyplot as plt
cities = ["Dhaka", "Chittagong", "Sylhet", "Khulna"]
population = [22, 9, 4, 3]   # মিলিয়নে

plt.bar(cities, population, color="green")
plt.title("city wise population")
plt.xlabel("city")
plt.ylabel("population in million")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

scores = np.random.normal(70, 10, 1000)   # ১০০০ জনের নম্বর, গড় ৭০

plt.hist(scores, bins=30, color="skyblue", edgecolor="black")
plt.title("নম্বরের Distribution")
plt.xlabel("নম্বর")
plt.ylabel("কতজন")
plt.show()
```
