import some data

```python
import pandas as pd
import numpy as np
data = {
    "Name": ["Rahim", "Karim", "Salma", "Nadia", "Karim"],
    "Age": [22, 25, np.nan, 23, 25],
    "Score": [78, 92, 85, np.nan, 92]
}

df = pd.DataFrame(data)
print(df)
```

Let's find out the missing values

```python
print(df.isnull().sum())
```

let's find out the duplicate value

```python
print(df.duplicated())
```
