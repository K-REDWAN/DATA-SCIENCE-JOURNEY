import some data

```python
import pandas as pd
import numpy as np
data = {
    "Name": ["Rahim", "Karim", "Salma", "Nadia", "Karim"],
    "Age": [22, 25, np.nan, 23, 25],
    "Score": [78, 92, 85, np.nan, 92]
}

df = pd.DataFrame(data)
print(df)
```

```python
print(df.dtypes)
```

Let's find out the missing values

```python
print(df.isnull().sum())
```

let's find out the duplicate value

```python
print(df.duplicated())
```

Let's delete the duplicate

```python
df_no_duplicates = df.drop_duplicates()
print(df_no_duplicates)
```

Let' handle the missing values
1. drop or delete the value

```python
print(df)

df_dropped = df.dropna()
print(df_dropped)
```
